### 1.2 Parse DateTime and Extract Features


```python
# Parse DateTime column (using format='mixed' to handle inconsistent datetime formats)
df['DateTime'] = pd.to_datetime(df['DateTime'], format='mixed')

# Extract temporal features
df['Date'] = df['DateTime'].dt.date
df['Year'] = df['DateTime'].dt.year
df['Month'] = df['DateTime'].dt.month
df['Day'] = df['DateTime'].dt.day
df['Hour'] = df['DateTime'].dt.hour
df['DayOfYear'] = df['DateTime'].dt.dayofyear
df['WeekOfYear'] = df['DateTime'].dt.isocalendar().week
df['Quarter'] = df['DateTime'].dt.quarter

# Map day of week to numeric (if not already)
dow_mapping = {
    'Monday': 0, 'Tuesday': 1, 'Wednesday': 2, 'Thursday': 3,
    'Friday': 4, 'Saturday': 5, 'Sunday': 6
}
df['DayOfWeek_Numeric'] = df['DoW'].map(dow_mapping)

print("Temporal features extracted successfully!")
print(f"\nDate range: {df['DateTime'].min()} to {df['DateTime'].max()}")
print(f"Total days covered: {(df['DateTime'].max() - df['DateTime'].min()).days}")
```

```python
# Column groupings for easier analysis
scheduled_cols = [col for col in df.columns if '예정' in col and '작년' not in col]
expected_cols = [col for col in df.columns if '예상' in col]
lastyear_cols = [col for col in df.columns if '작년' in col]
weather_cols = ['Temperature', 'Humidity']

print("Column groupings:")
print(f"\nScheduled (예정) columns: {scheduled_cols}")
print(f"\nExpected (예상) columns: {expected_cols}")
print(f"\nLast year (작년) columns: {lastyear_cols}")
print(f"\nWeather columns: {weather_cols}")
```

### 4.3 Product Segmentation


```python
# Segment products by volume and volatility
# Volume: Low (<25th percentile), Medium (25-75th), High (>75th)
# Volatility: Stable (CV <= 1), Moderate (1 < CV <= 2), Volatile (CV > 2)

volume_25 = product_stats['Mean_Orders'].quantile(0.25)
volume_75 = product_stats['Mean_Orders'].quantile(0.75)

def categorize_volume(val):
    if val < volume_25:
        return 'Low'
    elif val < volume_75:
        return 'Medium'
    else:
        return 'High'

def categorize_volatility(cv):
    if cv <= 1:
        return 'Stable'
    elif cv <= 2:
        return 'Moderate'
    else:
        return 'Volatile'

product_stats['Volume_Category'] = product_stats['Mean_Orders'].apply(categorize_volume)
product_stats['Volatility_Category'] = product_stats['CV'].apply(categorize_volatility)
product_stats['Segment'] = product_stats['Volume_Category'] + '_' + product_stats['Volatility_Category']

# Segment distribution
segment_dist = product_stats.groupby(['Volume_Category', 'Volatility_Category']).size().reset_index(name='Count')

print("Product Segmentation:")
display(segment_dist.pivot(index='Volume_Category', columns='Volatility_Category', values='Count').fillna(0).astype(int))

# Visualize segmentation
fig, ax = plt.subplots(figsize=(12, 8))
scatter = ax.scatter(product_stats['Mean_Orders'], product_stats['CV'], 
                     c=product_stats['Volume_Category'].astype('category').cat.codes,
                     s=product_stats['Record_Count'], alpha=0.6, cmap='viridis')
ax.axhline(y=1, color='green', linestyle='--', linewidth=1.5, alpha=0.7, label='CV = 1')
ax.axhline(y=2, color='red', linestyle='--', linewidth=1.5, alpha=0.7, label='CV = 2')
ax.axvline(x=volume_25, color='blue', linestyle='--', linewidth=1.5, alpha=0.7)
ax.axvline(x=volume_75, color='blue', linestyle='--', linewidth=1.5, alpha=0.7)
ax.set_xlabel('Mean Order Quantity', fontsize=12)
ax.set_ylabel('Coefficient of Variation (CV)', fontsize=12)
ax.set_title('Product Segmentation: Volume vs Volatility\n(bubble size = record count)', 
             fontsize=14, fontweight='bold')
ax.legend()
ax.grid(True, alpha=0.3)
ax.set_xlim(left=0)
ax.set_ylim(bottom=0)
plt.colorbar(scatter, ax=ax, label='Volume Category')
plt.tight_layout()
plt.show()
```

```python
# Outlier detection using IQR method
def detect_outliers_iqr(series):
    Q1 = series.quantile(0.25)
    Q3 = series.quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    outliers = ((series < lower_bound) | (series > upper_bound)).sum()
    return outliers, lower_bound, upper_bound

print("=" * 80)
print("OUTLIER ANALYSIS (IQR Method)")
print("=" * 80)

outlier_summary = []
for col in scheduled_cols + expected_cols:
    outliers, lower, upper = detect_outliers_iqr(df[col])
    outlier_summary.append({
        'Column': col,
        'Outlier_Count': outliers,
        'Outlier_Percentage': (outliers / len(df) * 100),
        'Lower_Bound': lower,
        'Upper_Bound': upper
    })

outlier_df = pd.DataFrame(outlier_summary)
display(outlier_df)
```
